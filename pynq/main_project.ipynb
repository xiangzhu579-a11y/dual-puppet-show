# K230 → PYNQ → 两块 RA6M5

| LED | 含义 |
|---|---|
| LED0 | 主循环每 0.5 秒翻转，闪烁表示持续监听 |
| LED1 | 收到 READY 后常亮 |
| LED2 | 两路 START 提交成功后亮；两路 STOP 成功后灭 |
| LED3 | 检测到非法指令、K230 接收错误或程序异常后亮，重启程序清除 |

正常退出时 LED0 灭，其余灯保留最后状态。LED2 不是 MCU 执行反馈；无 ACK 时不能证明人偶已启动或停止。
位流已更新为三 UART + 四位 LED GPIO（axi_gpio_0，0x41200000）。
READY 记录就绪且只提示一次；每条 START、STOP 都发送至两块 RA6M5，格式为 ASCII + CRLF。
第一块 RA6M5 RX 接 Pmod B Pin 2；第二块 RX 接 Pmod B Pin 4。两块板 GND 分别接 B5、B11。三路 UART 均为 115200、8N1。
程序使用 triple_uart.bit / triple_uart.hwh。先提交两路发送再等待 FIFO 排空，软件发送存在先后差，不保证严格同步。
